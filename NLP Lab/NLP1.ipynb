```python
from gtts import gTTS
import os
import time
import pyttsx3
```

```python
# Function to convert the text to speech
def textTtoSpeech(text, output_file="output.mp3", language="en"):
    # engine = pyttsx3.init()
    # start the counter
    start_time = time.perf_counter()
    # Select the language in which the speech in needed
    # to use offline i am using pyttsx3
    tts = gTTS(text=text, lang=language) 
    # Save the Output file
    tts.save(output_file)
    # End the timer
    end_time = time.perf_counter()

    # engine.runAndWait()

    # Calculate the conversion time
    conversion_time = end_time - start_time
    # Get the size of the output file saved in the folder where this notebook is stored
    file_size = os.path.getsize(output_file)

    # return the conversion time and file size
    return conversion_time, file_size
```

```python
# Function to count the number of words
def count_words(text):
    # split the input text into tokens, this will create list of words with comma separated
    # len counts the total number of words in the token list
    return len(text.split())
```

```python
def count_paragraphs(text):
    paragraphs = [p.strip() for p in text.split("\n\n") if p.strip()]
    return len(paragraphs)
```

```python
def static_text_input():
    text = """
     A convolutional network assumes that nearby pixels matter together. A recurrent model assumes that the past influences the present. 
     An attention model asks which parts of the input should interact.A graph neural network assumes that relations matter.
     An equivariant model assumes that some transformations should preserve meaning.
    """
    output_file = "static_output.mp3"
    conversion_time, file_size = textTtoSpeech(
        text,
        output_file
    )
    print("\n----- STATIC TEXT INPUT -----")
    print("Input text:")
    print(text)
    print("Number of words:", count_words(text))
    print("Conversion time:", round(conversion_time, 4), "seconds")
    print("Output audio filename:", output_file)
    print("Output audio file size:", file_size, "bytes")
```

```python
def dynamic_text_input():
    text = input("\nEnter the text to convert to speech: ")
    output_file = "dynamic_output.mp3"
    conversion_time, file_size = textTtoSpeech(
        text,
        output_file
    )

    print("\n----- DYNAMIC USER INPUT -----")
    print("Input text:", text)
    print("Number of words:", count_words(text))
    print("Conversion time:", round(conversion_time, 4), "seconds")
    print("Output audio filename:", output_file)
    print("Output audio file size:", file_size, "bytes")
```

```python
def text_file_input():
    filename = input("\nEnter text file name: ")
    if not os.path.exists(filename):
        print("File not found!")
        return
    with open(filename, "r", encoding="utf-8") as file:
        text = file.read()
    output_file = "file_output.mp3"
    conversion_time, file_size = textTtoSpeech(
        text,
        output_file
    )

    input_file_size = os.path.getsize(filename)
    print("\n----- TEXT FILE INPUT -----")
    print("Input file size:", input_file_size, "bytes")
    print("Number of words:", count_words(text))
    print("Number of paragraphs:", count_paragraphs(text))
    print("Conversion time:", round(conversion_time, 4), "seconds")
    print("Output audio filename:", output_file)
    print("Output audio file size:", file_size, "bytes")
```
