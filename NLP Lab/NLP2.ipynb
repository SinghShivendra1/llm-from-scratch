```python
# COMBINED CODE
text = 'IIIT Surat is a leading institute for computer science education'
# 1. Convert Text to lower case
lower_text = text.lower()
print('Lower text: '+str(lower_text))

# 2. Tokenize the text
tokens = lower_text.split()
print('tokens: '+str(tokens))

# 3. Get unique words
unique_words = set(tokens)
print('Unique words : '+str(unique_words)+' length = '+str(len(unique_words)))

# 4. Sort the word list
sorted_list = sorted(unique_words)
print('Sorted list: '+str(sorted_list))

# 5. Get the integer/position of the words
wordToIndex = {}
for index, word in enumerate(unique_words):
    wordToIndex[word]=index
print('Word to index: '+ str(wordToIndex))

# 6. create a vector of each word by marking its position as 1 and rest as 0
vectors = []
for word in tokens:
    vector = [0]*len(unique_words)
    vector[wordToIndex[word]] = 1
    vectors.append(vector)
print('Vectors: ')
print(vectors)
# 7: Create a matrix of the found vectors.
for row in vectors:
    print(row)
```

```python
from collections import defaultdict
import string

# Sample text data: sentences
corpus = [
    "IIIT Surat offers computer science education",
    "Surat offers computer engineering education."
]
# Function to preprocess text
def preprocess(text):
    # Convert to lowercase
    text = text.lower()
    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))
    # Tokenize: split the text into words
    tokens = text.split()
    return tokens

# Apply preprocessing to the sample corpus
processed_corpus = [preprocess(sentence) for sentence in corpus]
print(processed_corpus)
```

```python
# Initialize an empty set for the vocabulary
vocabulary = set()

# Build the vocabulary
for sentence in processed_corpus:
    vocabulary.update(sentence)

# Convert to a sorted list
vocabulary = sorted(list(vocabulary))
print("Vocabulary:", vocabulary)
```

```python
def create_bow_vector(sentence, vocab):
    vector = [0] * len(vocab)  # Initialize a vector of zeros
    for word in sentence:
        if word in vocab:
            idx = vocab.index(word)  # Find the index of the word in the vocabulary
            vector[idx] += 1  # Increment the count at that index
    return vector
```

```python
# Create BoW vector for each sentence in the processed corpus
bow_vectors = [create_bow_vector(sentence, vocabulary) for sentence in processed_corpus]
print("Bag of Words Vectors:")
for vector in bow_vectors:
    print(vector)
```
